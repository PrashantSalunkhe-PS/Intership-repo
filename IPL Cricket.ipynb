## Team-wise Winning Percentage

```python
wins = df['winner'].value_counts()

matches = pd.concat([df['team1'],df['team2']]).value_counts()

winning_percentage = pd.DataFrame({
    "Matches": matches,
    "Wins": wins
}).fillna(0)

winning_percentage["Winning %"] = (
    winning_percentage["Wins"] /
    winning_percentage["Matches"]*100
).round(2)

winning_percentage.sort_values("Winning %",ascending=False)
```

# Venue-wise Match Analysis

```python
venue_analysis = df.groupby("venue").agg(
    Matches=("venue","count"),
    Winners=("winner","nunique")
)
venue_analysis.sort_values("Matches",ascending=False)
```
