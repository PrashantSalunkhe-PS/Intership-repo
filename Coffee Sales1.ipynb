```python
df["date"] = pd.to_datetime(df["date"])
```

```python
df["Month"] = df["date"].dt.month_name()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df["money"], bins=20, kde=True)
plt.title("Distribution of Sales Amount")
plt.xlabel("Money")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(4,5))
sns.boxplot(x=df["money"])
plt.title("Boxplot of Sales Amount")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(
    y="coffee_name",
    data=df,
    order=df["coffee_name"].value_counts().index,
    palette="viridis"
)
plt.title("Coffee Sales Count")
plt.xlabel("Number of Orders")
plt.ylabel("Coffee Name")
plt.show()
```

```python
plt.figure(figsize=(4,5))
sns.countplot(
    x="cash_type",
    data=df,
    palette="Set2"
)
plt.title("Payment Method Distribution")
plt.xlabel("Payment Type")
plt.ylabel("Count")
plt.show()
```

```python
monthly_sales = df.groupby("Month")["money"].sum()

plt.figure(figsize=(12,6))
monthly_sales.plot(marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```
